```python
from getpass import getpass
from openai import OpenAI

client = OpenAI(
    base_url="https://uphold-suitor-scored.ngrok-free.dev/v1",
    api_key=getpass("Senha da API: "),
    timeout=600,
    max_retries=0,
)

print("Modelo:", client.models.list().data[0].id)

resposta = client.chat.completions.create(
    model="Qwen/Qwen2.5-Coder-14B-Instruct",
    temperature=0,
    max_completion_tokens=128,
    messages=[
        {"role": "user", "content": "Explique SQL injection em duas frases."}
    ],
)

print(resposta.choices[0].message.content)
print("Tokens:", resposta.usage)
```
